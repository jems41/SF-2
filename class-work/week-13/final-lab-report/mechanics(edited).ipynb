```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import math

df = pd.read_csv("clean_data_single_filter.csv", delimiter=",")

m = 0.00087  # Mass (kg)
A = 0.01916  # Cross-sectional area (m²) # 1.9022889
p = 1.204 # Air density (kg/m³)
g = 9.81  # Gravity (m/s²)

v_term = np.mean([-0.97035897, -1.02553846, -1.0791453, -0.99586325, 
                  -0.98684615, -1.02413675, -0.97470085, -0.82321368, 
                  -0.98632479, -0.95158974])
print(v_term)

C = (2 * m * g) / (p * A * (v_term ** 2))

def Acceleration(v):
    return -g + ((C * p * A * (v ** 2)) / (2 * m))

initial_position = 1.1  
initial_velocity = 0   
dt = 0.001  

time = [0]
position = [initial_position]
velocity = [initial_velocity]

while initial_position > 0:
    ay = Acceleration(initial_velocity)  
    new_velocity = initial_velocity + ay * dt  
    new_position = initial_position + new_velocity * dt

    if new_position < 0:
        new_position = 0  

    initial_position = new_position
    initial_velocity = new_velocity

    time.append(time[-1] + dt)
    position.append(new_position)
    velocity.append(new_velocity)

plt.figure(figsize=(8, 6)) 

plt.plot(time, position, label="Euler Method (Simulation)", linewidth=2, color='blue')

time_columns = [col for col in df.columns if "Time" in col]
position_columns = [col for col in df.columns if "Position" in col]

for t_col, p_col in zip(time_columns, position_columns):
    time_values = df[t_col].values
    position_values = df[p_col].values

    valid_indices = position_values > 0
    last_valid_index = valid_indices.nonzero()[0][-1]  

    plt.plot(time_values[:last_valid_index], position_values[:last_valid_index], linestyle="dashed", label=p_col, alpha=0.7)

plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Euler Simulation vs. Experimental Data")
plt.legend()  
plt.grid(True)  
plt.show()
```
