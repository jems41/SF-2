6. The data in `data_df` contains noisy data.  In this step you are to clean the noisy data (or extract the clean data). Your cleaned data **for each run**
- must not include the data that is after once the filter hits the motion sensor
- must include the curved data that transitions into a drop in position.  You should only keep about 0.5 to 1 second of the roughly constant position values before the drop of position takes place.  
Note that you should clean your data for each of the 10 runs and you should automate this process in your code with logical conditions, loops, etc.

7. Store your clean data into a new DataFrame called `clean_df`. 
8. Note that now that the values in the **Time** columns may have shifted due to the clean-up.  Update your `clean_df` to rescale all the **Time** columns to start the time from 0.0 in increments of 0.02. 
9. Write this to a new .csv file called `"clean_data_single_filter.csv"`. 

```python
def clean_and_smooth_data(time, position, run_num, cleaned_data_lst):
    valid_mask = position != 0
    time = time[valid_mask]
    position = position[valid_mask]

    difference = np.diff(position)
    slope_start = None
    for x in range(len(difference) - 2):
        if all(difference[x:(x + 3)] < 0) and slope_start is None:
            slope_start = x

    beforeslope_time= 0
    if slope_start > 0:
        beforeslope_time = max(np.where(time < (time[slope_start] - 1))[0], default =0)

    print(f"Run {run_num}: Before slope start index: ", beforeslope_time)

    cleaned_time = time[beforeslope_time:]
    cleaned_position = position[beforeslope_time:]
    smoothed_position = pd.Series(cleaned_position).rolling(window = 5).mean()
    smoothed_position.bfill(inplace = True)
    smoothed_position.ffill(inplace = True)

    for i in range(1, len(smoothed_position)):
        if smoothed_position.iloc[i] >smoothed_position.iloc[i -1]:  
            smoothed_position.iloc[i] = smoothed_position.iloc[i - 1]

    lowest_point = smoothed_position.idxmin()
    if lowest_point is None:
        lowest_point = len(smoothed_position) -1
        
    cleaned_time = cleaned_time[:(lowest_point + 1)]
    cleaned_position = smoothed_position[:(lowest_point + 1)]
    time_rescale = 0.02
    cleaned_time = np.arange(0,len(cleaned_time) * time_rescale, time_rescale)[: len(cleaned_time)]
    cleaned_df_run = pd.DataFrame({f"Time (s) Run #{run_num}": cleaned_time, f"Position (m) Run #{run_num}": cleaned_position})
    data_lst.append(cleaned_df_run)

    return cleaned_time, cleaned_position

data_lst = []
plot_data_lst = []

for run in range(10):
    time_column = f"Time (s) Run #{run + 1}"
    position_column = f"Position (m) Run #{run+ 1}"
    if time_column in data_df.columns and position_column in data_df.columns:
        time = data_df[time_column].values
        position = data_df[position_column].values
        cleaned_time, cleaned_position = clean_and_smooth_data(time, position, run +1,data_lst)
        
        plot_data_lst.append((cleaned_time, cleaned_position))

clean_df = pd.concat(data_lst, axis = 1)
clean_df.fillna(0, inplace = True)
clean_df.to_csv("clean_data_single_filter.csv", index = False)
```

10. Plot the data using `matplotlib` from all 10 runs as scatter plots on the same figure using your cleaned data.

11. Take the last 0.5 to 1 second of the x-values in your 10 runs, and draw a line of best fit on all your 10 graphs. 

```python
plt.figure(figsize = (5, 5))
for run, (cleaned_time, cleaned_position) in enumerate(plot_data_lst):
    plt.scatter(cleaned_time, cleaned_position, label = f"Run {run + 1}", s = 1)

    index = np.searchsorted(cleaned_time, cleaned_time[-1] - 0.5)
    time_fit, position_fit = cleaned_time[index:], cleaned_position[index:]
    if len(time_fit) > 1:
        coefficient = np.polyfit(time_fit, position_fit, 1)
        plt.plot(time_fit, np.polyval(coefficient, time_fit), linestyle = '-', linewidth = 2)

plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Cleaned Position vs Time for Each Run")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def clean_and_smooth_data(time, position, run_num, cleaned_data_lst):
    valid_mask = position != 0
    time = time[valid_mask]
    position = position[valid_mask]

    difference = np.diff(position)
    slope_start = None
    for x in range(len(difference) - 2):
        if all(difference[x:(x + 3)] < 0) and slope_start is None:
            slope_start = x

    beforeslope_time= 0
    if slope_start > 0:
        beforeslope_time = max(np.where(time < (time[slope_start] - 1))[0], default =0)

    print(f"Run {run_num}: Before slope start index: ", beforeslope_time)

    cleaned_time = time[beforeslope_time:]
    cleaned_position = position[beforeslope_time:]
    
    true_lowest_point = np.argmin(cleaned_position)
    smoothed_position = pd.Series(cleaned_position).rolling(window= 5).mean()
    smoothed = smoothed_position.iloc[:true_lowest_point]
    smoothed = smoothed.bfill()
    smoothed = smoothed.ffill()
    smoothed_position.iloc[:true_lowest_point] = smoothed

    smoothed_position.iloc[true_lowest_point:] = cleaned_position[true_lowest_point:]

    for i in range(1, len(smoothed_position)):
        if smoothed_position.iloc[i] >smoothed_position.iloc[i -1]:  
            smoothed_position.iloc[i] = smoothed_position.iloc[i - 1]

    lowest_point = smoothed_position.idxmin()
    if lowest_point is None:
        lowest_point = len(smoothed_position) -1
      
    cleaned_time = cleaned_time[:(lowest_point + 1)]
    cleaned_position = smoothed_position[:(lowest_point + 1)]
    time_rescale = 0.02
    cleaned_time = np.arange(0,len(cleaned_time) * time_rescale, time_rescale)[: len(cleaned_time)]
    cleaned_df_run = pd.DataFrame({f"Time (s) Run #{run_num}": cleaned_time, f"Position (m) Run #{run_num}": cleaned_position})
    data_lst.append(cleaned_df_run)

    return cleaned_time, cleaned_position

data_lst = []
plot_data_lst = []

for run in range(10):
    time_column = f"Time (s) Run #{run + 1}"
    position_column = f"Position (m) Run #{run+ 1}"
    if time_column in data_df.columns and position_column in data_df.columns:
        time = data_df[time_column].values
        position = data_df[position_column].values
        cleaned_time, cleaned_position = clean_and_smooth_data(time, position, run +1,data_lst)
        
        plot_data_lst.append((cleaned_time, cleaned_position))

clean_df = pd.concat(data_lst, axis = 1)
clean_df.fillna(0, inplace = True)
clean_df.to_csv("clean_data_multiple_filter.csv", index = False)
```
