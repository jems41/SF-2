# Lists Comprehension

```python
# given a list of lists, remove the empty list from it
# by keeping the original list as is
lst = [[1, 2], [], [3, 4], [], [5]]
result = []

for inner_list in lst:
    if inner_list:
        result.append(inner_list)
print(f'lst: {lst}')
print(f'result: {result}')

# write the same thing using a list comprehension
result2 = [inner_list for inner_list in lst if inner_list]
print(result2)

# remove the empty lists by modifying the original list of lists
for inner_list in lst:
    if not inner_list:
        lst.remove(inner_list)
print(lst)

result3 = [lst.remove(inner_list) for inner_list in lst if not inner_list]
print(result3)
```

# Sort a list of lists
- given a list of lists (assume integers), sort each sublist in decreasing order
- cannot use any built-in functions (no .sort)
- must define a sort function that allows to sort both in increasing or decreasing
- what is the time and space complexity?

```python
def insertion_sort(lst, mode):
    n = len(lst)

    if n <= 1:
        return lst
    
    for i in range(1, n):
        elem = lst[i]
        j = i - 1
        while j >= 0 and elem < lst[j]:
            lst[j+1] = lst [j]
            j -= 1
        lst[j+1] = elem

    if not mode:
        lst = lst[::-1]
    
    return lst
```

# List Example: transpose a matrix
- given an n by m matrix, write code to transpose it
- cannot use any built-in functions
-  [[1, 2]
    [3, 4]
    [5, 6]]
-  [[1, 3, 5],
    [2, 4, 5]]

```python
def transpose(lst_2d):
    n = len(lst_2d) # 3 row
    m = len(lst_2d[0]) # 2 col
    transposed = []

    for j in range(m):
        new_row = []
        for i in range(n):
            new_row.append(lst_2d[i][j])
        transposed.append(new_row)
    return transposed

print(transpose([[1, 2], [3, 4], [5, 6]]))
print(transpose([[1, 3, 5], [2, 4, 6]]))

lst_2d = [[1, 2], [3, 4], [5, 6]]
transposed = [[lst_2d[i][j] for i in range(len(lst_2d))] for j in range(len(lst_2d[0]))]

print(transposed)
```

```python
from math import factorial

def binom(n, k):
    if k > n or k < 0:
        return 0
    return factorial(n) // (factorial(n - k) * factorial(k))

def pascal(num):
    lst1 = []
    for i in range(num):
        lst2 = []
        for j in range(num):
            value = binom(i, j)
            if value != 0:
                lst2.append(value)
        lst1.append(lst2)
    return lst1
print(pascal(5))
```
