## filter + map practice

```python
def isPrime(n):
    if n < 3:
        return True
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

def makeComposite(n):
    if n < 3:
        return n + 3
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return n
    return n + 1

lst = [1, 3, 4, 6, 3, 5, 9, 56, 3, 235, 317]
lst2 = [1, 3, 4, 6, 3, 5, 9, 56, 3, 235, 317]
lst = filter(isPrime, lst)
print(list(lst))

lst3 = map(makeComposite, lst2)
print(list(lst3))
```
