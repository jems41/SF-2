# PLOTTING **POSITION VS. TIME GRAPHS** FOR EACH RUN

## Original Run 1:

```python
capstone_data_time = capstone_data["Time (s) Run #1"]
capstone_data_position = capstone_data["Position (m) Run #1"]
plt.figure(figsize=(10, 6))
plt.plot(capstone_data_time, capstone_data_position, label='Position', color='blue')
plt.title('Position vs. Time (Run #1)(ORIGINAL)')
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
plt.legend()
plt.grid(True)
plt.show()
```

### Plot For Capstone_Data

```python
for i in range(1, 11):
    time_column = f"Time (s) Run #{i}"
    position_column = f"Position (m) Run #{i}"
    capstone_data[[time_column, position_column]] = capstone_data[[time_column, position_column]].fillna(method = "ffill").fillna(method = "bfill")
    capstone_data.sort_values(by = time_column, ignore_index = True, inplace = True)

    if i == 2:
        cutoff_index = capstone_data[capstone_data[time_column] <= 0.54].index.max()
    else:
        min_index = capstone_data[position_column].argmin()
        diffs = capstone_data[position_column].diff().fillna(0)
        std_dev = diffs.std()
        first_increase = next((idx for idx, diff in enumerate(diffs) if diff > std_dev), len(diffs) - 1)
        cutoff_index = min(min_index, first_increase) if first_increase > min_index else min_index

    filtered_capstone_data = capstone_data.loc[:cutoff_index].copy()
    smoothed_positions = []

    for idx in range(len(filtered_capstone_data)):
        start = max(0, idx - 2)
        end = idx + 1
        window = filtered_capstone_data[position_column].iloc[start:end]
        smoothed_positions.append(window.mean())

    filtered_capstone_data["position_smooth"] = smoothed_positions
    endpoint = filtered_capstone_data.loc[cutoff_index, position_column]
    adjustment_range = 3

    for j in range(adjustment_range):
        index = cutoff_index - adjustment_range + 1 + j
        weight = (j / (adjustment_range - 1)) ** 2
        filtered_capstone_data.loc[index, "position_smooth"] = ((1 - weight) * filtered_capstone_data.loc[index, "position_smooth"] + weight * endpoint)
    
    filtered_capstone_data.set_index(time_column)[[position_column, "position_smooth"]].rename(columns={position_column: "Original", "position_smooth": "Smooth"}).plot(figsize = (8, 5), color = ["gray", "blue"], alpha = 0.8, grid = True, title = f"Run #{i}: Position vs. Time")
    plt.show()
```

### Plot For Capstone_Data_2:

```python
for i in range(1, 11): 
    time_column = f"Time (s) Run #{i}"
    position_column = f"Position (m) Run #{i}"

    capstone_data_2[[time_column, position_column]] = (capstone_data_2[[time_column, position_column]].fillna(method = "ffill").fillna(method = "bfill"))
    capstone_data_2.sort_values(by=time_column, ignore_index=True, inplace=True)

    diffs = capstone_data_2[position_column].diff().fillna(0)
    downward_start = next((idx for idx, diff in enumerate(diffs) if diffs[idx:idx+3].lt(0).all()), 0)
    std_dev = diffs.std()
    cutoff_index = next((idx for idx, diff in enumerate(diffs) if diff > std_dev), len(diffs) - 1) - 1
    cutoff_index = max(downward_start, cutoff_index)

    filtered_data = capstone_data_2.loc[downward_start:cutoff_index, [time_column, position_column]].copy()

    smoothed_data = filtered_data[position_column]
    for x in range(3):
        smoothed_data = smoothed_data.rolling(window = 5, center = True).mean().fillna(method = "bfill").fillna(method = "ffill")

    filtered_data["position_smooth"] = smoothed_data

    ax = filtered_data.plot(x = time_column, y = position_column, figsize = (8, 5), color = "gray", alpha = 0.8, grid=True, label = "Original", title = f"Run #{i}: Position vs. Time")
    filtered_data.plot(x = time_column, y = "position_smooth", ax=ax, color = "blue", alpha = 1, label = "Smoothed")
    plt.xlabel("Time (s)")
    plt.ylabel("Position (m)")
    plt.legend()
    plt.show()
```

# PROPER PLOTTING/CLEANING DATA FUNCTION

```python
# Function to clean and smooth data
def clean_and_smooth_data(data):
    cleaned_data = []
    data_copy = data.copy()

    for i in range(1, 11):
        time_col = f"Time (s) Run #{i}"
        position_col = f"Position (m) Run #{i}"

        data_copy[[time_col, position_col]] = data_copy[[time_col, position_col]].fillna(method="ffill").fillna(method="bfill")
        sorted_data = data_copy.sort_values(by=time_col, ignore_index=True)
        
        position_diff = sorted_data[position_col].diff().fillna(0)
        
        downward_start = 0
        for index_num in range(len(position_diff) - 2):
            if position_diff[index_num:index_num + 3].lt(0).all():  # Check if the next three differences are negative
                downward_start = index_num
                break

        standard_deviation = position_diff.std()
        cutoff_index = len(position_diff) - 1
        for index_num, diff in enumerate(position_diff):
            if diff > standard_deviation:
                cutoff_index = index_num
                break
        cutoff_index = max(downward_start, cutoff_index - 1)

        filtered_data = sorted_data.loc[downward_start:cutoff_index, [time_col, position_col]].copy()

        smoothed_position = filtered_data[position_col]
        for smoothed_amount in range(3):
            smoothed_position = smoothed_position.rolling(window=5, center=True).mean().fillna(method="bfill").fillna(method="ffill")
        filtered_data["smoothed_position"] = smoothed_position

        min_position_idx = filtered_data[position_col].idxmin()
        cleaned_data.append(filtered_data.loc[:min_position_idx].copy())

    return cleaned_data

cleaned_capstone_data = clean_and_smooth_data(capstone_data)
cleaned_capstone_data_2 = clean_and_smooth_data(capstone_data_2)

for dataset, title in zip([cleaned_capstone_data, cleaned_capstone_data_2], ["Capstone Data", "Capstone Data 2"]):
    for run_num, filtered_data in enumerate(dataset, 1):
        time_col = f"Time (s) Run #{run_num}"
        position_col = f"Position (m) Run #{run_num}"
        plt.figure(figsize=(8, 5))
        plt.plot(filtered_data[time_col], filtered_data[position_col], color="gray", alpha=0.6, label="Original")
        plt.plot(filtered_data[time_col], filtered_data["smoothed_position"], color="blue", label="Smoothed")
        plt.title(f"{title} - Run #{run_num}")
        plt.xlabel("Time (s)")
        plt.ylabel("Position (m)")
        plt.grid(True)
        plt.legend()
        plt.show()
```
